# NVL72-class engineering demonstration
Run from the project root or notebook directory after installation. Input provenance is in `data/sources.yaml`; all results depend on assumed branch hydraulics.

```python
from pathlib import Path
import os
if Path.cwd().name == "notebooks": os.chdir("..")
from nvl72.config import load_config
from nvl72.solver import solve
import pandas as pd
baseline = solve(load_config())
pd.Series(baseline["metrics"])
```

```python
trays = pd.DataFrame(baseline["trays"]).set_index("tray_id")
trays[["actual_flow_LPM", "target_flow_LPM"]].plot.bar(figsize=(12,4), ylabel="Flow at supply density [L/min]")
```

```python
candidate = solve(load_config("config/optimized_example.yaml"))
from nvl72.reporting import comparison
pd.DataFrame(comparison(baseline, candidate))
```
